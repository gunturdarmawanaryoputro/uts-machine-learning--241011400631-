### Soal A3 – Training dengan Normal Equation
1. Latih model `LinearRegression` (normal equation, seperti di materi) menggunakan data training.
2. Tampilkan nilai **intercept** ($b$) dan **koefisien** ($m$), lalu tuliskan persamaan garis regresinya: $\hat{y} = b + mX$.
3. **Interpretasikan** arti nilai $m$ dalam konteks kasus ini (misal: "setiap kenaikan biaya iklan 1 juta Rp, penjualan naik sebesar ...").

````python
# 1. Kelas LinearRegression menggunakan Normal Equation
class LinearRegressionNormal:
    def fit(self, X, y):
        # Tambahkan kolom intercept (bias) bernilai 1 ke dalam matriks X
        X_b = np.c_[np.ones((X.shape[0], 1)), X]
        # Normal Equation: theta = (X^T * X)^(-1) * X^T * y
        theta = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
        self.intercept_ = theta[0]
        self.coef_ = theta[1:]
        
    def predict(self, X):
        X_b = np.c_[np.ones((X.shape[0], 1)), X]
        return X_b.dot(np.r_[self.intercept_, self.coef_])

# Latih model menggunakan data training (X_train, y_train)
model_lr = LinearRegressionNormal()
model_lr.fit(X_train, y_train)

# 2. Tampilkan nilai intercept (b) dan koefisien (m) serta persamaan garis regresi
print(f"Intercept (b)  : {model_lr.intercept_:.4f}")
print(f"Koefisien (m)  : {model_lr.coef_[0]:.4f}")
print(f"Persamaan Garis: y_hat = {model_lr.intercept_:.4f} + {model_lr.coef_[0]:.4f} * X")
````

### Soal A4 – Evaluasi Model
Pada materi belum ada evaluasi untuk model regresi. **Implementasikan sendiri** fungsi-fungsi berikut menggunakan NumPy:

$$
MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i - \hat{y}_i| \qquad
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2 \qquad
RMSE = \sqrt{MSE}
$$

$$
R^2 = 1 - \frac{\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}{\sum_{i=1}^{n}(y_i - \bar{y})^2}
$$

1. Hitung MAE, MSE, RMSE, dan $R^2$ pada **data training** dan **data testing**.
2. Tampilkan hasilnya dalam satu tabel (boleh pakai `pd.DataFrame`).
3. Jelaskan arti nilai RMSE dan $R^2$ yang Anda dapatkan. Apakah model ini sudah baik?

````python
# Jawaban A4# Implementasi fungsi evaluasi regresi menggunakan NumPy
def evaluate_regression(y_true, y_pred):
    mae = np.mean(np.abs(y_true - y_pred))
    mse = np.mean((y_true - y_pred) ** 2)
    rmse = np.sqrt(mse)
    
    ss_total = np.sum((y_true - np.mean(y_true)) ** 2)
    ss_residual = np.sum((y_true - y_pred) ** 2)
    r2 = 1 - (ss_residual / ss_total)
    
    return {'MAE': mae, 'MSE': mse, 'RMSE': rmse, 'R²': r2}

# 1. Hitung pada data training dan data testing
y_train_pred = model_lr.predict(X_train)
y_test_pred = model_lr.predict(X_test)

train_metrics = evaluate_regression(y_train, y_train_pred)
test_metrics = evaluate_regression(y_test, y_test_pred)

# 2. Tampilkan hasilnya dalam satu tabel DataFrame
df_eval = pd.DataFrame([train_metrics, test_metrics], index=['Training', 'Testing'])
display(df_eval)
````

### Soal A5 – Gradient Descent & Learning Rate
1. Latih model `LinearRegressionGD` (dari materi) dengan `n_iterations=5000` menggunakan **3 nilai learning rate**: `0.001`, `0.02`, dan `0.1`.
2. Untuk setiap learning rate, tampilkan intercept dan koefisien akhirnya, lalu bandingkan dengan hasil normal equation (Soal A3).
3. Buat grafik **loss (MSE) vs iterasi** untuk ketiga learning rate (boleh dalam satu grafik atau *subplot*; gunakan skala log jika perlu).
4. Jelaskan apa yang terjadi pada masing-masing learning rate: mana yang **terlalu lambat**, mana yang **pas**, dan mana yang **gagal konvergen (divergen)**? Mengapa bisa terjadi?

> *Tips: jika muncul `RuntimeWarning: overflow`, itu adalah bagian dari hasil eksperimen. Jelaskan artinya.*

````python
# 1. Definisi kelas LinearRegressionGD
class LinearRegressionGD:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.lr = learning_rate
        self.n_iters = n_iterations
        self.loss_history = []
        
    def fit(self, X, y):
        n_samples = X.shape[0]
        X_b = np.c_[np.ones((n_samples, 1)), X]
        self.theta = np.zeros(X_b.shape[1])
        
        for _ in range(self.n_iters):
            y_pred = X_b.dot(self.theta)
            mse = np.mean((y_pred - y) ** 2)
            self.loss_history.append(mse)
            
            # Hitung gradien
            gradients = (2/n_samples) * X_b.T.dot(y_pred - y)
            # Update parameter theta
            self.theta -= self.lr * gradients
            
        self.intercept_ = self.theta[0]
        self.coef_ = self.theta[1:]

# 2. Latih model dengan 3 nilai learning rate berbeda
lrs = [0.001, 0.02, 0.1]
plt.figure(figsize=(9, 5))

for lr in lrs:
    gd_model = LinearRegressionGD(learning_rate=lr, n_iterations=5000)
    gd_model.fit(X_train, y_train)
    
    print(f"Learning Rate = {lr}")
    print(f"  -> Intercept: {gd_model.intercept_:.4f}")
    print(f"  -> Koefisien: {gd_model.coef_[0]:.4f}")
    print(f"  -> Bandingkan dengan Normal Equation (Intercept: {model_lr.intercept_:.4f}, Koefisien: {model_lr.coef_[0]:.4f})\n")
    
    # 3. Plot grafik loss (MSE) vs iterasi
    plt.plot(gd_model.loss_history, label=f'LR = {lr}')

plt.yscale('log')
plt.title('Grafik Loss (MSE) vs Iterasi untuk Berbagai Learning Rate')
plt.xlabel('Jumlah Iterasi')
plt.ylabel('MSE (Skala Log)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
````

### Soal B2 – Split Data & Training
1. Bagi data dengan `test_size=0.25` dan `random_state=RANDOM_STATE`.
2. Latih model `LogisticRegression` (gradient descent, dari materi) dengan `learning_rate=0.1` dan `n_iterations=10000`.
3. Tampilkan intercept ($b$) dan koefisien ($m_1, m_2$), lalu tuliskan persamaannya:
   $$P(\text{Spam}) = \sigma(b + m_1 \cdot \text{JumlahLink} + m_2 \cdot \text{KataPromo})$$
4. Buat grafik **log loss vs iterasi**. Apakah loss sudah turun dan stabil?

````python
# 1. Bagi data dengan test_size=0.25 dan random_state=RANDOM_STATE
X_train_e, X_test_e, y_train_e, y_test_e = train_test_split(X_email, y_email, test_size=0.25, random_state=RANDOM_STATE)

# 2. Definisi dan Training Model Logistic Regression (Gradient Descent)
class LogisticRegressionCustom:
    def __init__(self, learning_rate=0.1, n_iterations=10000):
        self.lr = learning_rate
        self.n_iters = n_iterations
        self.loss_history = []
        
    def sigmoid(self, z):
        return 1 / (1 + np.exp(-np.clip(z, -500, 500)))
    
    def fit(self, X, y):
        n_samples, n_features = X.shape
        X_b = np.c_[np.ones((n_samples, 1)), X]
        self.theta = np.zeros(n_features + 1)
        
        for _ in range(self.n_iters):
            z = X_b.dot(self.theta)
            y_pred = self.sigmoid(z)
            
            # Hitung Log Loss
            epsilon = 1e-15
            y_pred_clipped = np.clip(y_pred, epsilon, 1 - epsilon)
            loss = -np.mean(y * np.log(y_pred_clipped) + (1 - y) * np.log(1 - y_pred_clipped))
            self.loss_history.append(loss)
            
            # Gradient Descent
            gradient = (1 / n_samples) * X_b.T.dot(y_pred - y)
            self.theta -= self.lr * gradient
            
        self.intercept_ = self.theta[0]
        self.coef_ = self.theta[1:]

# Latih model
model_log = LogisticRegressionCustom(learning_rate=0.1, n_iterations=10000)
model_log.fit(X_train_e, y_train_e)

# 3. Tampilkan intercept (b) dan koefisien (m1, m2) serta persamaan
print(f"Intercept (b)      : {model_log.intercept_:.4f}")
print(f"Koefisien (m1, m2) : {model_log.coef_}")
print(f"Persamaan          : P(Spam) = sigma({model_log.intercept_:.4f} + {model_log.coef_[0]:.4f} * JumlahLink + {model_log.coef_[1]:.4f} * KataPromo)")

# 4. Buat grafik log loss vs iterasi
plt.figure(figsize=(8, 5))
plt.plot(model_log.loss_history, color='purple')
plt.title('Grafik Log Loss vs Iterasi (Logistic Regression)')
plt.xlabel('Iterasi')
plt.ylabel('Log Loss')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
````

### Soal B3 – Evaluasi Model
Selain `accuracy_score` dan `confusion_matrix` dari materi, **implementasikan sendiri** fungsi berikut:

$$
Precision = \frac{TP}{TP + FP} \qquad
Recall = \frac{TP}{TP + FN} \qquad
F1 = 2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}
$$

1. Hitung accuracy, confusion matrix, precision, recall, dan F1-score pada **data testing**.
2. Jelaskan arti nilai **FP** (*false positive*) dan **FN** (*false negative*) pada kasus filter spam. Kesalahan mana yang menurut Anda **lebih merugikan pengguna email**? Mengapa?

````python
# Fungsi implementasi metrik evaluasi sendiri menggunakan NumPy
def evaluate_classification(y_true, y_pred):
    tp = np.sum((y_true == 1) & (y_pred == 1))
    tn = np.sum((y_true == 0) & (y_pred == 0))
    fp = np.sum((y_true == 0) & (y_pred == 1))
    fn = np.sum((y_true == 1) & (y_pred == 0))
    
    # Confusion Matrix dalam bentuk array 2x2
    conf_matrix = np.array([[tn, fp], [fn, tp]])
    
    # Metrik
    accuracy = (tp + tn) / len(y_true)
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    
    return {
        'Confusion Matrix': conf_matrix,
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1,
        'TP': tp, 'TN': tn, 'FP': fp, 'FN': fn
    }

# 1. Hitung pada data testing
y_test_pred = model_log.predict(X_test_e, threshold=0.5)
eval_results = evaluate_classification(y_test_e, y_test_pred)

print("--- Hasil Evaluasi Data Testing ---")
print(f"Accuracy  : {eval_results['Accuracy']:.4f}")
print(f"Precision : {eval_results['Precision']:.4f}")
print(f"Recall    : {eval_results['Recall']:.4f}")
print(f"F1-Score  : {eval_results['F1-Score']:.4f}")
print("\nConfusion Matrix:")
print(eval_results['Confusion Matrix'])
````

### Soal B4 – Decision Boundary
1. Hitung persamaan **decision boundary** (garis saat probabilitas = 0.5):
   $$b + m_1 x_1 + m_2 x_2 = 0 \;\Rightarrow\; x_2 = -\frac{b + m_1 x_1}{m_2}$$
2. Buat visualisasi data **testing** beserta *background* probabilitas (`contourf`) dan garis decision boundary, seperti di materi.
3. Tunjukkan (beri tanda/penjelasan) data testing mana saja yang **salah diklasifikasikan**.

````python
# 1. Visualisasi Data Testing beserta Contourf (Background Probabilitas) dan Decision Boundary
plt.figure(figsize=(9, 6))

# Membuat grid untuk background contourf
x1_min, x1_max = X_email[:, 0].min() - 1, X_email[:, 0].max() + 1
x2_min, x2_max = X_email[:, 1].min() - 1, X_email[:, 1].max() + 1
xx1, xx2 = np.meshgrid(np.linspace(x1_min, x1_max, 200),
                       np.linspace(x2_min, x2_max, 200))

grid_points = np.c_[xx1.ravel(), xx2.ravel()]
probs = model_log.predict_proba(grid_points).reshape(xx1.shape)

# Plot contourf untuk background probabilitas
plt.contourf(xx1, xx2, probs, levels=50, cmap='RdBu', alpha=0.4)
plt.colorbar(label='Probabilitas Spam')

# Plot data testing
y_test_pred = model_log.predict(X_test_e, threshold=0.5)
misclassified = (y_test_e != y_test_pred)

# Data testing benar
plt.scatter(X_test_e[y_test_e == 0 & ~misclassified, 0], X_test_e[y_test_e == 0 & ~misclassified, 1], color='blue', marker='o', label='Bukan Spam (Benar)', s=50)
plt.scatter(X_test_e[y_test_e == 1 & ~misclassified, 0], X_test_e[y_test_e == 1 & ~misclassified, 1], color='red', marker='o', label='Spam (Benar)', s=50)

# Data testing yang SALAH diklasifikasikan (diberi tanda khusus, misal marker 'X' hitam/kuning)
if np.any(misclassified):
    plt.scatter(X_test_e[misclassified, 0], X_test_e[misclassified, 1], color='black', marker='X', s=120, label='Salah Diklasifikasikan')

# 2. Plot Decision Boundary (garis saat probabilitas = 0.5)
# Rumus: x2 = -(b + m1 * x1) / m2
b = model_log.intercept_
m1, m2 = model_log.coef_[0], model_log.coef_[1]
x1_boundary = np.array([x1_min, x1_max])
x2_boundary = -(b + m1 * x1_boundary) / m2

plt.plot(x1_boundary, x2_boundary, color='green', linewidth=2.5, linestyle='--', label='Decision Boundary (p=0.5)')

plt.title('Decision Boundary & Background Probabilitas (Data Testing)')
plt.xlabel('Jumlah Link')
plt.ylabel('Jumlah Kata Promo')
plt.legend(loc='upper right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
````

### Soal B5 – Eksperimen Threshold
Secara default, model memprediksi Spam jika $P(\text{Spam}) \geq 0.5$.

1. Gunakan `predict_proba` lalu buat prediksi dengan threshold **0.3**, **0.5**, dan **0.7**.
2. Untuk setiap threshold, hitung confusion matrix, precision, dan recall. Tampilkan dalam satu tabel.
3. Jelaskan pengaruh menaikkan/menurunkan threshold terhadap precision dan recall. Berdasarkan jawaban Anda di Soal B3, **threshold mana yang Anda pilih** untuk filter spam? Beri alasan.

````python
# 1. Eksperimen Threshold (0.3, 0.5, dan 0.7)
thresholds = [0.3, 0.5, 0.7]
results_list = []

for th in thresholds:
    # Prediksi menggunakan threshold tertentu
    preds_th = model_log.predict(X_test_e, threshold=th)
    
    # Hitung komponen confusion matrix
    tp = np.sum((y_test_e == 1) & (preds_th == 1))
    tn = np.sum((y_test_e == 0) & (preds_th == 0))
    fp = np.sum((y_test_e == 0) & (preds_th == 1))
    fn = np.sum((y_test_e == 1) & (preds_th == 0))
    
    # Hitung Precision dan Recall
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    accuracy = (tp + tn) / len(y_test_e)
    
    results_list.append({
        'Threshold': th,
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'TP': tp, 'TN': tn, 'FP': fp, 'FN': fn
    })

# 2. Tampilkan hasil dalam satu tabel DataFrame
df_thresholds = pd.DataFrame(results_list).set_index('Threshold')
display(df_thresholds[['Accuracy', 'Precision', 'Recall', 'TP', 'TN', 'FP', 'FN']])
````
